### Load the CSV files

```python
from pathlib import Path
import pandas as pd

# --- Robust base directory resolution ---
try:
    BASE_DIR = Path(__file__).resolve().parent   # .py script
except NameError:
    BASE_DIR = Path.cwd()                        # Jupyter notebook

DATA_DIR = BASE_DIR / "ch03_csv"
assert DATA_DIR.exists(), f"Missing folder: {DATA_DIR}"

print("Using DATA_DIR:", DATA_DIR.resolve())

csv_files = {
    "es_futures_quote": "es_futures_quote.csv",
    "es_futures_specs": "es_futures_specs.csv",
    "es_options_specs": "es_options_specs.csv",
    "es_options_chain": "es_options_chain_synthetic.csv",
    "quikstrike_inputs": "quikstrike_inputs_sample.csv",
    "es_expirations": "es_expirations.csv",
    "es_vol_observations": "es_vol_observations.csv",
    "spx_options_chain": "spx_options_chain_sample.csv",
    "xsp_options_chain": "xsp_options_chain_sample.csv",
    "spx_specs": "spx_specs.csv",
}

dfs = {}
missing = []

for k, fname in csv_files.items():
    path = DATA_DIR / fname
    if path.exists():
        dfs[k] = pd.read_csv(path)
    else:
        missing.append(fname)

if missing:
    print("WARNING: Missing CSV files:")
    for f in missing:
        print("  -", f)

summary = pd.DataFrame(
    [{"dataset": k, "rows": v.shape[0], "cols": v.shape[1]} for k, v in dfs.items()]
)

print(summary)
```

### DataFrames
